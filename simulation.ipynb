```
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider

# 1. Setup Simulation Data
np.random.seed(42)
sampling_rate = 250
duration_dream = 10
duration_present = 2

time_dream = np.linspace(0, duration_dream, sampling_rate * duration_dream)
time_present = np.linspace(0, duration_present, sampling_rate * duration_present)

# Dream String (Infinite Noise) & Present String (Sensory Query)
rem_wave = np.sin(2 * np.pi * 20 * time_dream) + np.sin(2 * np.pi * 40 * time_dream)
infinite_dream_string = rem_wave + np.random.normal(0, 1.5, len(time_dream))

waking_wave = np.sin(2 * np.pi * 18 * time_present)
present_string = waking_wave + np.random.normal(0, 0.5, len(time_present))

# Inject Hidden Match at the 5-second mark
target_idx = 1250
match_length = 150
infinite_dream_string[target_idx : target_idx + match_length] = present_string[100 : 100 + match_length]

# 2. Build Interactive Plot UI
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 5))
plt.subplots_adjust(bottom=0.25)

# Plot Base Data
ax1.plot(time_dream, infinite_dream_string, color='gray', alpha=0.6, label='Deep Dream Memory')
ax2.plot(time_present, present_string, color='blue', label='Present Reality String')

# Interactive Dragging States
current_x = 1.0
span = ax1.axvspan(current_x, current_x + 0.6, color='yellow', alpha=0.4, label='Scanning Mind...')

ax1.set_title("Deep Neural Database (REM Sleep Memory String)")
ax1.set_ylabel("Amplitude (µV)")
ax1.grid(True, alpha=0.3)
ax1.legend(loc='upper right')

ax2.set_title("Present Waking Processing (Shorter Reality String)")
ax2.set_xlabel("Time (Seconds)")
ax2.set_ylabel("Amplitude (µV)")
ax2.grid(True, alpha=0.3)

# 3. Add Slider Framework
ax_slider = plt.axes([0.15, 0.08, 0.7, 0.03])
scan_slider = Slider(ax_slider, 'Scan Timeline', 0.0, 9.4, valinit=current_x)

def update(val):
    pos = scan_slider.val
    target_time = 5.0
    
    # Your Theory Logic: Snapping/Locking Mechanism
    if abs(pos - target_time) < 0.15:
        pos = target_time
        span.set_xy([[pos, -5], [pos, 5], [pos + 0.6, 5], [pos + 0.6, -5], [pos, -5]])
        span.set_color('red')
        span.set_alpha(0.5)
    else:
        span.set_xy([[pos, -5], [pos, 5], [pos + 0.6, 5], [pos + 0.6, -5], [pos, -5]])
        span.set_color('yellow')
        span.set_alpha(0.4)
        
    fig.canvas.draw_idle()

scan_slider.on_changed(update)
plt.show()
```
